```python
def astar(graph, heuristic, start, goal):
    open_list = [(heuristic[start], 0, start, [start])]
    visited = set()

    while open_list:
        open_list.sort()

        f, g, node, path = open_list.pop(0)

        if node == goal:
            print("Path:", path)
            print("Cost:", g)
            return

        if node in visited:
            continue

        visited.add(node)

        for neighbour, cost in graph[node].items():
            if neighbour not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbour]

                open_list.append((
                    new_f, new_g, neighbour,
                    path + [neighbour]
                ))

    print("Path not found")


graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'C': 2, 'D': 5},
    'C': {'A': 4, 'B': 2, 'D': 1},
    'D': {'B': 5, 'C': 1}
}

heuristic = {
    'A': 3,
    'B': 2,
    'C': 1,
    'D': 0
}

astar(graph, heuristic, 'A', 'D')
```
