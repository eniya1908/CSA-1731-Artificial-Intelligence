```python
from collections import deque

N = 8

def print_board(board):
    for row in board:
        print(" ".join(row))
    print()

def is_safe(board, row, col):
    for i in range(row):
        if board[i][col] == 'Q':
            return False

        if col - (row - i) >= 0:
            if board[i][col - (row - i)] == 'Q':
                return False

        if col + (row - i) < N:
            if board[i][col + (row - i)] == 'Q':
                return False

    return True

def solve(board, row):
    if row == N:
        print_board(board)
        return True

    for col in range(N):
        if is_safe(board, row, col):
            board[row][col] = 'Q'

            if solve(board, row + 1):
                return True

            board[row][col] = '.'

    return False

board = [['.' for _ in range(N)] for _ in range(N)]

if not solve(board, 0):
    print("Solution does not exist")
```
