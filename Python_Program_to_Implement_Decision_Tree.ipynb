```python
from sklearn.tree import DecisionTreeClassifier

# Training data
X = [
    [1, 1],
    [1, 0],
    [0, 1],
    [0, 0]
]

# Output labels
y = [1, 1, 0, 0]

# Create Decision Tree
model = DecisionTreeClassifier()

# Train model
model.fit(X, y)

# Predict output
prediction = model.predict([[1, 0]])

print("Prediction:", prediction[0])
```
