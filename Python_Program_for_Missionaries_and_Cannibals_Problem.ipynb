```python
from collections import deque

def solve():
    start = (3, 3, 1)
    goal = (0, 0, 0)

    queue = deque([(start, [])])
    visited = set()

    moves = [(1, 0), (2, 0), (0, 1),
             (0, 2), (1, 1)]

    while queue:
        state, path = queue.popleft()

        if state == goal:
            for step in path + [state]:
                print(step)
            return

        if state in visited:
            continue

        visited.add(state)

        m, c, boat = state

        for x, y in moves:
            if boat == 1:
                new = (m-x, c-y, 0)
            else:
                new = (m+x, c+y, 1)

            a, b, side = new

            if 0 <= a <= 3 and 0 <= b <= 3:
                if (a == 0 or a >= b) and \
                   (3-a == 0 or 3-a >= 3-b):

                    if new not in visited:
                        queue.append((new, path + [state]))

solve()
```
