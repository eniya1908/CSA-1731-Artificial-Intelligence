```python
import numpy as np
from sklearn.neural_network import MLPClassifier

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Output data (XOR)
y = np.array([0, 1, 1, 0])

# Create neural network
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='relu',
    solver='lbfgs',
    max_iter=2000,
    random_state=1
)

# Train the network
model.fit(X, y)

# Predict output
prediction = model.predict([[1, 1]])

print("Predicted Output:", prediction[0])
```
